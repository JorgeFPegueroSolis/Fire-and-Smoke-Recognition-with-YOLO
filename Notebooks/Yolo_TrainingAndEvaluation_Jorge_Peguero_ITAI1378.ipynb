```python
model = YOLO('yolov8n-cls.pt')
#Defining the model
```

```python
print(len(list(model.model.parameters()))) #Verifying the model downloaded correctly and figuring out the number of parameters.
```

```python
model.train(data ='/content/the_wildfire_dataset_2n_version',epochs = 20, patience = 5, freeze = 5, cls_pw = 0.25, save_period = 1, project ='/content/drive/MyDrive/CV_Project', name='C' )

#training the model and saving its results in Google Drive
```

```python
best_model = YOLO('/content/drive/MyDrive/CV_Project/CV_Project/weights/best.pt') #Defining the best model from the previous training, which is stored in Google Drive
```

```python
evaluation = best_model.val(data ='/content/the_wildfire_dataset_2n_version', split = 'test', verbose =True, project ='/content/drive/MyDrive/CV_Project', name='CV_Project_evaluation'  )

#performing evaluation on the test split and saving its results on Google Drive
```
